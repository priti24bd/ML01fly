## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print("Why ML beats a fixed rule:")
print()
print("A simple fixed rule might rank products using only one factor, such as the number of previous clicks.")
print()
print("However, product relevance can depend on multiple factors and their combinations. With many users, searches, and products, writing a separate if-statement for every situation would be difficult and inflexible.")
print()
print("Machine learning can learn patterns from historical data and combine multiple signals to produce a more flexible ranking.")
print()
print("I would still compare the ML approach with a simple baseline rule to check whether ML actually improves ranking quality.")
```
