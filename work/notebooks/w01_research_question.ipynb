## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

* **The Core Question:** Can we predict which products a user is most likely to click or buy based on their search terms?
* **The Unit of Analysis:** A single user search session.
* **The Output:** A ranked list of product IDs sorted by relevance.
* **The Decision:** Deciding the display order of items on the search results page.
* **The Action:** The frontend website automatically promotes high-scoring items to the top slots.
* **The Cost of a Wrong Call:** False positives show irrelevant junk, driving users away. False negatives bury great products on deep pages, losing immediate revenue.
* **Why ML is needed:** Humans cannot write manual rules for millions of unique search and user combinations. ML scales to find hidden interaction patterns automatically.
